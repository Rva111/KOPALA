# Flame / Freefall — start the app

Use your **Python 3.11** kernel. Keep this notebook in the extracted app folder. Run Cell 1 once; run Cell 2 to open the white chat interface. If prompted, select your existing private `.env.local` file. No earlier notebooks are required.

```python
import sys, subprocess
from pathlib import Path
app_folder = Path.cwd()
if not (app_folder / "requirements.txt").is_file():
    raise RuntimeError("Open this notebook from the extracted Flame_Freefall_App folder.")
subprocess.check_call([sys.executable, "-m", "pip", "install", "-r", str(app_folder / "requirements.txt")])
```

```python
import os, sys, subprocess, webbrowser
from pathlib import Path
app_folder = Path.cwd()
sys.path.insert(0, str(app_folder))
from assistant import api_key
try:
    api_key()
except RuntimeError:
    import tkinter as tk
    from tkinter import filedialog
    root = tk.Tk()
    root.withdraw()
    root.attributes("-topmost", True)
    selected = filedialog.askopenfilename(title="Select your existing private .env.local file")
    root.destroy()
    if selected:
        os.environ["FLAME_ENV_FILE"] = selected
if "flame_process" in globals() and flame_process.poll() is None:
    print("The app is already running: http://127.0.0.1:8765")
    webbrowser.open("http://127.0.0.1:8765")
else:
    flame_process = subprocess.Popen([sys.executable, "launch.py"], cwd=app_folder)
    print("Opening http://127.0.0.1:8765 — leave this kernel running.")
```

To stop the app, run the optional cell below. If the launcher reports port 8765 is occupied, close the earlier app window first.

```python
if "flame_process" in globals() and flame_process.poll() is None:
    flame_process.terminate()
    print("App stopped.")
```
